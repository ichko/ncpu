```python
W = 100
H = 100
r = 20
small_r = 10
spacing = (55, 30)
margin = 20
dataset = NCPUDataset(
    W=W,
    H=H,
    small_r=small_r,
    r=r,
    spacing=spacing,
    margin=margin,
    sampler=sample_XOR_gate,
)

n = 10
fig, axes = plt.subplots(2, n, figsize=(n * 2, 4))
# Take only 10 samples from the infinite generator
for i, (input_img, output_img) in enumerate(itertools.islice(dataset, 10)):
    # Display input in the top row
    axes[0, i].imshow(input_img)
    axes[0, i].axis("off")  # Hide axes

    # Display output in the bottom row
    axes[1, i].imshow(output_img)
    axes[1, i].axis("off")

plt.tight_layout()
plt.show()
```
